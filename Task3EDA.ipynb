```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("/content/SHOP_2015-05-21.csv")
```

```python
print(df.head())
print(df.shape)
print(df.info())
```

```python
print(df.isnull().sum())
```

```python
df=df.drop_duplicates()
```

```python
df["date"]=pd.to_datetime(
    df["date"],
    format="mixed",
    errors="coerce"
)
```

```python
df.info()
```

```python
df=df.sort_values("date")
```

```python
df=df.set_index("date")
```

```python
df=df.dropna()
```

```python
df.info()
```

```python
df["Daily_Price_Change"]=df["close"]-df["open"]
```

```python
df["Daily_Return_%"]=((df["close"]-df["open"])/df["open"])*100
```

```python
df["Price_Range"]=df["high"]-df["low"]
```

```python
print(df.head())
```

```python
df.describe()
```

```python
print("Mean Return:",
      df["Daily_Return_%"].var())
```

```python
print("Return Variance:",
      df["Daily_Return_%"].var())
```

```python
print("Return Standard Deviation:",
      df["Daily_Return_%"].std())
```

```python
plt.figure(figsize=(12,5))
plt.plot(df.index,df["volume"])
plt.title("Shopify Trading Volume Trend")
plt.xlabel("Date")
plt.ylabel("Volume")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(10,5))
plt.hist(df["Daily_Return_%"],bins=30)
plt.title("Shopify Daily Return Distribution")
plt.xlabel("Index")
plt.ylabel("Price Range")
plt.show()
```

```python
plt.figure(figsize=(12,5))
plt.plot(df.index,df.Price_Range)
plt.title("Shopify Price Range Trend")
plt.xlabel("Index")
plt.ylabel("Price Range")
plt.xticks(rotation=45)
plt.show()
```
